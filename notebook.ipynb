1) The main difference between regression and classification is that regression predicts a continuous numerical value, whereas classification predicts a class or category label.   

2) A confusion table is a grid that compares actual class labels against a model's predicted class labels. It helps us understand performance by showing where the model is making correct predictions and where it is confusing one class for another.   

3) The SSE quantifies the total squared difference between actual values and predicted values, measuring a regression model's overall fit and error level.   

4) Overfitting happens when a model learns training data and noise too closely, performing poorly on new data, while underfitting happens when a model is too simple to learn the underlying patterns in the data at all.   

5) Splitting data into training and test sets ensures that hyperparameter k is selected based on how well the model generalizes to unseen data rather than just memorizing the training set.   

6) Predicting a class label provides a direct answer, but it loses information about how confident the model is in that decision. Predicting a probability distribution preserves uncertainty and confidence levels, though it requires an extra step to choose a threshold for taking action.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# 1. Load data and perform initial EDA
df = pd.read_csv("data/land_mines.csv")

print("--- EDA Summary ---")
print("Head:\n", df.head())
print("\nShape:", df.shape)
print("\nData Info:")
df.info()
print("\nSummary Statistics:\n", df.describe())
print("\nMissing Values:\n", df.isnull().sum())
print("\nTarget Class Distribution (mine_type):\n", df["mine_type"].value_counts())

# 2. Split data 50/50 into training and validation sets
X = df[["voltage", "height", "soil"]]
y = df["mine_type"]

X_train_raw, X_valid_raw, y_train, y_valid = train_test_split(
    X, y, test_size=0.50, random_state=65
)

# Scale features using MinMaxScaler
scaler = MinMaxScaler()
X_train_scaled = scaler.fit_transform(X_train_raw)
X_valid_scaled = scaler.transform(X_valid_raw)

# 3. Select optimal k by evaluating model performance across candidates
k_values = range(1, 21)
accuracies = []

for k in k_values:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train_scaled, y_train)
    acc = knn.score(X_valid_scaled, y_valid)
    accuracies.append(acc)
    print(f"k = {k:2d} | Validation Accuracy: {acc:.4f}")

# Find best k
best_k = k_values[accuracies.index(max(accuracies))]
print(f"\nSelected optimal k: {best_k} (Accuracy: {max(accuracies):.4f})")

# Fit optimal model
knn_optimal = KNeighborsClassifier(n_neighbors=best_k)
knn_optimal.fit(X_train_scaled, y_train)
y_pred = knn_optimal.predict(X_valid_scaled)

# 4. Print overall accuracy and confusion matrix
overall_acc = accuracy_score(y_valid, y_pred)
cm = confusion_matrix(y_valid, y_pred)

print("\n--- Model Evaluation ---")
print(f"Overall Accuracy on Validation Set: {overall_acc:.2%}\n")
print("Confusion Matrix (Rows: Actual, Columns: Predicted):")
cm_df = pd.DataFrame(
    cm, 
    index=[f"Actual {i}" for i in range(1, 6)], 
    columns=[f"Pred {i}" for i in range(1, 6)]
)
print(cm_df)

print("\nDetailed Classification Report:")
print(classification_report(y_valid, y_pred))
```
